```python
df_estados = df[df['place_type'] == 'state']
df['date'] = pd.to_datetime(df['date']) 
```

```python
df_estados = df[df['place_type'] == 'state'].copy()
```

```python
df_estados.head()
```

```python
df_estados.isnull().sum()
```

======================

Agora vamos selecionar e inserir no dataframe novo apenas as colunas que irei utilizar

======================

```python

df_estados = df_estados[['state', 'date', 'new_confirmed', 'new_deaths', 'last_available_confirmed', 'last_available_deaths', 'last_available_confirmed_per_100k_inhabitants']]
```

```python
df_estados.head()
```

```python
df_estados = df_estados.copy()
```

```python
df_estados['regiao'] = df_estados['state'].map({
    'SP': 'Sudeste',
    'RJ': 'Sudeste',
    'MG': 'Sudeste',
    'ES': 'Sudeste',
    'AM': 'Norte', 
    'RR': 'Norte', 
    'AP': 'Norte', 
    'PA': 'Norte', 
    'TO': 'Norte', 
    'RO': 'Norte', 
    'AC': 'Norte', 
    'MA': 'Nordeste',
    'PI': 'Nordeste',
    'CE': 'Nordeste',
    'RN': 'Nordeste',
    'PB': 'Nordeste',
    'PE': 'Nordeste',
    'AL': 'Nordeste',
    'SE': 'Nordeste',
    'BA': 'Nordeste',
    'MT': 'Centro-Oeste',
    'MS': 'Centro-Oeste',
    'GO': 'Centro-Oeste',
    'DF': 'Centro-Oeste',
    'PR': 'Sul',
    'SC': 'Sul',
    'RS': 'Sul',
})

#Norte: AM, RR, AP, PA, TO, RO, AC
#Nordeste: MA, PI, CE, RN, PB, PE, AL, SE, BA
#Centro-Oeste: MT, MS, GO, DF
#Sudeste: SP, RJ, MG, ES
#Sul: PR, SC, RS
```

```python
df_estados.dtypes
```

```python
df_estados['regiao'].value_counts()
```

```python
df_estados.head()
```

```python
df_estados['taxa_obitos'] = df_estados['last_available_deaths'] / df_estados['last_available_confirmed']
```

```python
df_estados[['state', 'date', 'taxa_obitos']].tail(10)
```

```python
df_estados.sample(10)
```

```python
casos_por_dia = df_estados.groupby('date')['new_confirmed'].sum()
print(casos_por_dia.tail(10))
```

```python
casos_por_dia.plot(kind='line', figsize=(12, 4))
plt.title('Casos confirmados')
plt.ylabel('Número de casos')
plt.xlabel('Tempo')

plt.show()
```

```python
casos_por_estado = df_estados.groupby('regiao')['last_available_confirmed_per_100k_inhabitants'].max()
casos_por_estado.plot(kind='bar', figsize=(12, 4))
plt.title('Casos confirmados')
plt.ylabel('Taxa de casos confirmados')
plt.xlabel('Regiões')

plt.show()
```

==========================

Treinamento do modelo

==========================

```python
df_estados.sample(5)
```

```python
evolucao_casos = df_estados.groupby(['date','regiao'])['new_confirmed'].sum()
```

```python
evolucao_casos.head(5)
```

```python
evolucao_casos = df_estados.groupby(['date','regiao'])['new_confirmed'].sum().unstack()
```

```python
evolucao_casos.sample(5)
```

```python
evolucao_casos.plot(kind='line', figsize=(12, 4))
plt.title('Evolução dos casos')
plt.ylabel('Taxa de casos confirmados')
plt.xlabel('Regiões')
plt.show()
```
